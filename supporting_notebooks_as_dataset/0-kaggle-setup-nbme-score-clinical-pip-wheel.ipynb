you attach to the competition notebook (internet OFF).

|--------|------|
| Qwen3.5-9B weights | `/kaggle/working/models/qwen_35_9b/` |

| pip wheels | `/kaggle/working/wheels/` |

> **After committing**: Kaggle → Output → "Use as Dataset" → attach to competition notebook.

```python
import subprocess
import sys
from pathlib import Path

WHEELS_DIR = Path("/kaggle/working/wheels")
WHEELS_DIR.mkdir(parents=True, exist_ok=True)

PACKAGES = [
    "transformers>=5.5.0",
    "peft>=0.15.0",
    "bitsandbytes>=0.45.0",
    "accelerate>=1.0.0",
    "rapidfuzz>=3.0.0",
]

# ----------------------------
# Step 1: Download wheels ONCE
# ----------------------------
print("▶ Downloading pip wheels ...")

subprocess.check_call([
    sys.executable, "-m", "pip", "download",
    "--dest", str(WHEELS_DIR),
    *PACKAGES,
])

wheels = list(WHEELS_DIR.glob("*.whl"))
print(f"✓ {len(wheels)} wheels downloaded → {WHEELS_DIR}")

# ----------------------------
# Step 2: Extract versions from filenames
# ----------------------------
def parse_wheel_name(wheel_path):
    name = wheel_path.name
    parts = name.split("-")
    package = parts[0]
    version = parts[1]
    return package, version

print("\n▶ Extracted versions from downloaded wheels:")
versions = {}

for wheel in wheels:
    pkg, ver = parse_wheel_name(wheel)
    versions[pkg.lower()] = ver
    print(f"{pkg}: {ver}")

# ----------------------------
# Step 3: Optional - verify target packages
# ----------------------------
print("\n▶ Requested package versions (resolved):")
for req in PACKAGES:
    pkg_name = req.split(">=")[0].lower()
    resolved = versions.get(pkg_name, "NOT FOUND")
    print(f"{pkg_name}: {resolved}")
```

```python
import os

print("▶ Disk usage summary:")
result = subprocess.run(["du", "-sh", str(WHEELS_DIR)],
                        capture_output=True, text=True)
print(result.stdout)

print("\n▶ Wheel directory contents:")
for w in sorted(WHEELS_DIR.glob("*.whl")):
    print(f"  {w.name}")

print("\n✓ Setup complete. Commit this notebook to save outputs as Kaggle dataset.")
```
